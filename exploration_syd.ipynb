```python
eda_cols = [
    'Solar_Irradiance',
    'Temperature',
    'Humidity',
    'Wind_Speed'
]

df[eda_cols].describe()
```

```python


df[eda_cols].hist(
    figsize=(12, 8),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
for col in eda_cols:
    plt.figure(figsize=(6, 4))
    plt.boxplot(df[col].dropna())
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)
    plt.show()
```
